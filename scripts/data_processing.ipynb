### Data Deduplication

```python
import json

dataset_root = "affine-dataset/openspiel"
input_path = "clobber_determine.jsonl"
output_path = "clobber_determine_dedup.jsonl"
key = "conversation"   # property to deduplicate on

seen = set()

with open(dataset_root + "/" + input_path, "r", encoding="utf-8") as fin, \
     open(dataset_root + "/" + output_path, "w", encoding="utf-8") as fout:

    for line in fin:
        obj = json.loads(line)
        value = obj.get(key)
        
        value_str = json.dumps(
            value,
            sort_keys=True,        # canonical order for dicts
            ensure_ascii=False,
            separators=(",", ":")  # remove whitespace differences
        )

        if value_str not in seen:
            seen.add(value_str)
            fout.write(json.dumps(obj, ensure_ascii=False) + "\n")
```
